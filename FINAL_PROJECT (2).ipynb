# **STEP-5 : DEFINE INPUT FEATURES(X) AND TARGET(Y)**

```python


# Select input features
features = [
    'Make',
    'Model',
    'Vehicle Class',
    'Engine Size(L)',
    'Cylinders',
    'Transmission',
    'Fuel Type'
]

# Define input and target
X = df[features]
y = df['CO2 Emissions(g/km)']

# Display input features
print("Input Features (X):")
display(X.head())

# Display target variable
print("\nTarget Variable (y):")
display(y.head())

# Check shapes
print("\nShape of X:", X.shape)
print("Shape of y:", y.shape)

# Check data types
print("\nInput Feature Data Types:")
print(X.dtypes)

# Check target data type
print("\nTarget Data Type:", y.dtype)
```

# **STEP-6 : SPILT THE DATASET INTO TRAINING AND TESTING SETS**

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display dataset sizes
print("Training and Testing Data Split")
print("--------------------------------")
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))
```

# **STEP-7 : ENCODING AND PREPROCESSING**

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# Define categorical and numerical features
categorical_features = [
    'Make',
    'Model',
    'Vehicle Class',
    'Transmission',
    'Fuel Type'
]

numerical_features = [
    'Engine Size(L)',
    'Cylinders'
]

# Preprocessing for Linear Regression
linear_preprocessor = ColumnTransformer(
    transformers=[
        ('categorical',
         OneHotEncoder(handle_unknown='ignore'),
         categorical_features),

        ('numerical',
         StandardScaler(),
         numerical_features)
    ]
)

# Preprocessing for Random Forest
rf_preprocessor = ColumnTransformer(
    transformers=[
        ('categorical',
         OneHotEncoder(handle_unknown='ignore'),
         categorical_features),

        ('numerical',
         'passthrough',
         numerical_features)
    ]
)

print("Preprocessing pipelines created successfully!")

print("\nCategorical Features:", categorical_features)
print("Numerical Features:", numerical_features)
print("\nLinear Regression: Encoding + Standardization")
print("Random Forest: Encoding + Original Numerical Values")
```

# **STEP-8 : TRAIN LINEAR REGRESSION & RANDOM FOREST REGRESSOR**

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

# Create Linear Regression pipeline
linear_model = Pipeline([
    ('preprocessing', linear_preprocessor),
    ('model', LinearRegression())
])

# Create Random Forest pipeline
rf_model = Pipeline([
    ('preprocessing', rf_preprocessor),
    ('model', RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ))
])

# Train Linear Regression
linear_model.fit(X_train, y_train)
print("Linear Regression trained successfully!")

# Train Random Forest Regressor
rf_model.fit(X_train, y_train)
print("Random Forest Regressor trained successfully!")
```

# **STEP-9 : EVALUATE MODEL PERFORMANCE**

```python
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)
import numpy as np
import pandas as pd

# Make predictions
y_pred_linear = linear_model.predict(X_test)
y_pred_rf = rf_model.predict(X_test)

# Function to calculate regression metrics
def evaluate_model(y_true, y_pred, model_name):
    return {
        'Model': model_name,
        'R2 Score': r2_score(y_true, y_pred),
        'MAE': mean_absolute_error(y_true, y_pred),
        'MSE': mean_squared_error(y_true, y_pred),
        'RMSE': np.sqrt(mean_squared_error(y_true, y_pred))
    }

# Evaluate both models
results = pd.DataFrame([
    evaluate_model(y_test, y_pred_linear, 'Linear Regression'),
    evaluate_model(y_test, y_pred_rf, 'Random Forest Regressor')
])

# Display results
print("Model Performance Comparison:")
display(results.round(4))
```

# **STEP-10 : CROSS-VALIDATION & HYPERPARAMETER TUNING**

PERFORM 5 FOLD CROSS VALIDATION

```python
from sklearn.model_selection import KFold, cross_val_score

# Create 5-fold cross-validation
kfold = KFold(n_splits=5, shuffle=True, random_state=42)

# Evaluate Random Forest using R2 score
cv_scores = cross_val_score(
    rf_model,
    X_train,
    y_train,
    cv=kfold,
    scoring='r2',
    n_jobs=-1
)

print("Cross-Validation R2 Scores:", cv_scores)
print("Mean CV R2 Score:", cv_scores.mean())
print("Standard Deviation:", cv_scores.std())
```

TUNE RANDOM FOREST HYPERPARAMRTER

```python
from sklearn.model_selection import RandomizedSearchCV

# Define hyperparameter search space
param_dist = {
    'model__n_estimators': [100, 200, 300],
    'model__max_depth': [None, 10, 20, 30],
    'model__min_samples_split': [2, 5, 10],
    'model__min_samples_leaf': [1, 2, 4],
    'model__max_features': [1.0, 'sqrt']
}

# Create randomized search
random_search = RandomizedSearchCV(
    estimator=rf_model,
    param_distributions=param_dist,
    n_iter=15,
    scoring='r2',
    cv=kfold,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

# Fit using training data only
random_search.fit(X_train, y_train)

# Display best parameters and score
print("Best Hyperparameters:")
print(random_search.best_params_)

print("\nBest Cross-Validation R2 Score:")
print(random_search.best_score_)

# Save the best tuned model
best_rf_model = random_search.best_estimator_
```

# **STEP-11 : COMPARE MODEL & SELECT BEST MODEL**

```python
# Compare All Models

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)
import numpy as np
import pandas as pd

# Predictions from all three models
y_pred_linear = linear_model.predict(X_test)
y_pred_rf = rf_model.predict(X_test)
y_pred_tuned_rf = best_rf_model.predict(X_test)

# Function to calculate performance metrics
def evaluate_model(y_true, y_pred, model_name):
    mse = mean_squared_error(y_true, y_pred)

    return {
        'Model': model_name,
        'R2 Score': r2_score(y_true, y_pred),
        'MAE': mean_absolute_error(y_true, y_pred),
        'MSE': mse,
        'RMSE': np.sqrt(mse)
    }

# Create comparison table
comparison = pd.DataFrame([
    evaluate_model(y_test, y_pred_linear, 'Linear Regression'),
    evaluate_model(y_test, y_pred_rf, 'Random Forest'),
    evaluate_model(y_test, y_pred_tuned_rf, 'Tuned Random Forest')
])

# Display comparison
display(comparison.round(4))

# Select the model with the lowest test MAE
best_index = comparison['MAE'].idxmin()
best_model_name = comparison.loc[best_index, 'Model']

model_map = {
    'Linear Regression': linear_model,
    'Random Forest': rf_model,
    'Tuned Random Forest': best_rf_model
}

best_model = model_map[best_model_name]

print("Selected Model:", best_model_name)
print("Selection criterion: Lowest test MAE")
```

# **STEP-12 : PLOT ACTUAL vs PREDICTED EMISSION & ANALYSE RESIDUALS**

```python


import matplotlib.pyplot as plt
import seaborn as sns

# Predictions from selected model
y_pred_best = best_model.predict(X_test)

# Plot 1: Actual vs Predicted
plt.figure(figsize=(7, 6))

sns.scatterplot(x=y_test, y=y_pred_best, alpha=0.5)

# Ideal prediction line
lower = min(y_test.min(), y_pred_best.min())
upper = max(y_test.max(), y_pred_best.max())

plt.plot([lower, upper], [lower, upper], 'r--')

plt.xlabel('Actual CO2 Emissions (g/km)')
plt.ylabel('Predicted CO2 Emissions (g/km)')
plt.title(f'Actual vs Predicted - {best_model_name}')
plt.tight_layout()
plt.show()

# Calculate residuals
residuals = y_test - y_pred_best

# Plot 2: Residuals vs Predicted
plt.figure(figsize=(7, 5))

sns.scatterplot(x=y_pred_best, y=residuals, alpha=0.5)
plt.axhline(y=0, color='red', linestyle='--')

plt.xlabel('Predicted CO2 Emissions (g/km)')
plt.ylabel('Residuals (g/km)')
plt.title('Residual Analysis')
plt.tight_layout()
plt.show()

# Print residual summary
print("Mean Residual:", residuals.mean())
print("Residual Standard Deviation:", residuals.std())
```
